```python
import scipy.stats as stats 
import pandas as pd 
  
# loading the csv file 
data = [2.1, 1.6, 2.8, 0.4, 3.6, 1.7, 3.3, 2.6, 1.1, 1.9, 3.6, 3.2, 2.3, 2.1, 2.8]
  
# perform one sample t-test 
t_statistic, p_value = stats.ttest_1samp(a=data, popmean=1.9) 
print(t_statistic , p_value)
```

```python
import scipy.stats as stats

fuel = [2.1, 1.6, 2.8, 0.4, 3.6, 1.7, 3.3, 2.6, 1.1, 1.9, 3.6, 3.2, 2.3, 2.1, 2.8]
t_stat, p_two_tailed = stats.ttest_1samp(fuel, 1.9)

# En-sidigt test (mu > 1.9), så dela p-värdet på 2
p_one_tailed = p_two_tailed / 2
print(p_one_tailed)
```

```python
import pandas as pd
import scipy.stats as stats

# Skapa korrekt DataFrame
observed = pd.DataFrame(
    [
        [25, 40, 30, 50],
        [15, 20, 25, 30],
        [10, 15, 20, 25]
    ],
    index=['Satisfied', 'Neutral', 'Dissatisfied'],
    columns=['Less Than 1 Year', '1-3 Years', '3-5 Years', 'Over 5 Years']
)

# Chi-square test
chi2, p, dof, expected = stats.chi2_contingency(observed)

# Gör expected till DataFrame
expected_df = pd.DataFrame(expected, index=observed.index, columns=observed.columns)

# Total antal observationer
total = observed.values.sum()
relative_freq = expected_df / total

print("Expected Values:")
display(expected_df.round(2))

print("Expected Relative Frequencies (i.e. Probability Distribution):")
display(relative_freq.round(4))

# Resultat
print(f"\nChi-Square Statistic = {chi2:.4f}")
print(f"p-value = {p}")
```

```python
import scipy.stats as stats

# Datan
no_physicalAct = [23,26,51,49,58,37,29,44]
physicalAct_1hour = [22,27,39,39,46,48,49,65]
physicalAct_3hour = [59,66,38,49,56,60,56,62]

# Utför Kruskal-Wallis-test
h_statistic, p_value = stats.kruskal(no_physicalAct, physicalAct_1hour, physicalAct_3hour)

print(f"H-statistic: {h_statistic}")
print(f"P-value: {p_value}")
```

```python
import scipy.stats as stats

# Datan från bilden
with_beer = [12.4, 15.5, 17.9, 9.7, 19.6, 16.5, 15.1, 16.3, 13.3]
without_beer = [10.0, 14.2, 18.0, 10.1, 14.2, 12.1, 15.1, 12.4, 12.7]

# Utför paired t-test
t_statistic, p_value = stats.ttest_rel(with_beer, without_beer)

print(f"T-statistic: {t_statistic}")
print(f"P-value: {p_value}")
```

```python
t_stat_welch, p_value_unequal_var = stats.ttest_ind(with_beer, without_beer, equal_var=False)
```
