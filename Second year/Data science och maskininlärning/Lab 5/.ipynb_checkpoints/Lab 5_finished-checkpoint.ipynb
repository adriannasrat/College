## Labb 5 – Task 1 (Sentimentanalys med hjälp av polaritet)
**Uppgift:** Utföra sentimentanalys med hjälp av NLP på verified_reviews i amazon_alexa.tsv  

1. Rensa text (ta bort symboler, siffror etc.)
2. Tokenisera, lowercase
3. Lemmatisera
4. Ta bort stopwords
5. Visa topp 20 frekventa ord
6. Beräkna polarity och visa antal positiva/negativa recensioner


```python
# Steg 1
import pandas as pd
import re
import nltk
import matplotlib.pyplot as plt
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from collections import Counter
from textblob import TextBlob
```

```python
# Steg 2:a
df = pd.read_csv("data/amazon_alexa.tsv", sep="\t")
df.head()
```

```python
# Steg 2:b
df['cleaned'] = df['verified_reviews'].astype(str)
df['cleaned'] = df['cleaned'].apply(lambda x: re.sub(r'[^A-Za-z\s]', '', x)).str.lower()

df['cleaned'].head()
```

```python
# Steg 2:c , 2:d
lemmatizer = WordNetLemmatizer()
stop_words = set(stopwords.words('english'))

def preprocess(text):
    tokens = nltk.word_tokenize(text.lower())  # lowercase + tokenisera
    lemmatized = [lemmatizer.lemmatize(word) for word in tokens if word not in stop_words]
    return lemmatized

df['tokens'] = df['cleaned'].apply(preprocess)
df['tokens'].head()
```

```python
#steg 3
all_words = [word for tokens in df['tokens'] for word in tokens]
word_freq = Counter(all_words)
top_words = word_freq.most_common(20)

words, freqs = zip(*top_words)
plt.figure(figsize=(10, 5))
plt.bar(words, freqs)
plt.xticks(rotation=45)
plt.title("Top 20 Most Frequent Words")
plt.tight_layout()
plt.show()
```

```python
# Steg 4
df['polarity'] = df['verified_reviews'].apply(lambda x: TextBlob(str(x)).sentiment.polarity)
df['sentiment'] = df['polarity'].apply(lambda x: 'positive' if x > 0 else 'negative')
df[['verified_reviews', 'polarity', 'sentiment']].head()
```

```python
# Steg 4
plt.figure(figsize=(6, 4))
df['sentiment'].value_counts().plot(kind='bar', color=['green', 'red'])
plt.title("Sentiment Polarity: Positive vs Negative")
plt.ylabel("Number of Reviews")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## Labb 5 – Task 2 (Textklassificering med hjälp av Randomforest)
***Uppgift:*** Utföra textklassificering med hjälp av Random Forest på "review" i moviereviews.tsv  

1. Rensa text (ta bort symboler, siffror etc.)
2. Tokenisera, lowercase
3. Lemmatisera
4. Ta bort stopwords
5. Visa topp 20 frekventa ord
6. Dela upp dataset 70/30
7. Träna Random Forest-modell med Pipeline
8. Mäta accuracy


```python
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv("data/moviereviews.tsv", sep="\t")

# Ta bort rader där recensionen saknas
df.dropna(subset=['review'], inplace=True)

df.head()
```

```python
# Skapa en ny kolumn med rensad text
df['cleaned'] = df['review'].astype(str)
df['cleaned'] = df['cleaned'].apply(lambda x: re.sub(r'[^A-Za-z\s]', '', x))
df['cleaned'].head()
```

```python
lemmatizer = WordNetLemmatizer()
stop_words = set(stopwords.words('english'))

def preprocess(text):
    tokens = nltk.word_tokenize(text.lower()) 
    lemmatized = [lemmatizer.lemmatize(word) for word in tokens if word not in stop_words]
    return lemmatized

df['tokens'] = df['cleaned'].apply(preprocess)
df['tokens'].head()
```

```python
all_words = [word for tokens in df['tokens'] for word in tokens]
word_freq = Counter(all_words)
top_words = word_freq.most_common(20)

words, freqs = zip(*top_words)
plt.figure(figsize=(10, 5))
plt.bar(words, freqs)
plt.xticks(rotation=45)
plt.title("Top 20 Most Frequent Words")
plt.tight_layout()
plt.show()
```

```python
# Vi använder originalrecensionerna som textinput
X = df['review']
y = df['label']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
```

```python
model = Pipeline([
    ('vectorizer', CountVectorizer()),
    ('classifier', RandomForestClassifier(n_estimators=100, random_state=42))
])

# Tränar vi modellen
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)

# Utvärdera modellens noggrannhet
accuracy = accuracy_score(y_test, y_pred)
print(f"Modellens accuracy: {accuracy:.2%}")
```
