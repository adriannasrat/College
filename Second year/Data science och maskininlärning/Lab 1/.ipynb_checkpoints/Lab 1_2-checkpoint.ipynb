### Skillnaden mellan z-test och t-test

När man utför hypotesprövning stöter man ofta på **z-testet** och **t-testet**. Båda används för att jämföra medelvärden, men det finns en viktig skillnad mellan dem:

#### Grundläggande skillnad:
$$
\text{Z-test används när populationens standardavvikelse } \sigma \text{ är känd.}
$$  
$$
\text{T-test används när populationens standardavvikelse } \sigma \text{ är okänd, och man istället använder } s \text{ från stickprovet.}
$$

---

#### Z-test:
- Används vid stora stickprov ($n \geq 30$) och när **populationens standardavvikelse ($\sigma$) är känd**.
- Fördelningen är normalfördelningen (z-fördelningen).
- Formeln för teststatistikan är:

---

#### T-test:
- Används vid små stickprov ($n < 30$) och när **$\sigma$ är okänd**.
- Man använder istället **stickprovets standardavvikelse ($s$)**.
- T-fördelningen används, som liknar normalfördelningen men med "tjockare svansar" (för att kompensera för osäkerhet vid små stickprov).

---

#### Sammanfattning:

| Egenskap               | Z-test                     | T-test                      |
|------------------------|----------------------------|-----------------------------|
| Känd standardavvikelse | Ja ($\sigma$ känd)         | Nej ($\sigma$ okänd)     |
| Stickprovsstorlek      | Stort ($n \geq 30$)        | Litet ($n < 30$ är vanligt) |
| Fördelning             | Normalfördelning (Z)       | T-fördelning                |

### T-statistic

Formeln för t-statistikan i ett ensidigt eller tvåsidigt t-test är:

$$
t = \frac{\bar{x} - \mu}{\frac{s}{\sqrt{n}}}
$$

Där:
- $\bar{x}$ = stickprovsmedelvärde  
- $\mu$ = hypotetiskt populationsmedelvärde (under nollhypotesen)  
- $s$ = stickprovets standardavvikelse  
- $n$ = stickprovsstorlek

```python
myDataset = [125.2994, 126.4402, 171.2639, 143.5338,
160.2288, 97.2396, 132.6253, 125.3658, 157.1991, 155.7616, 130.6605, 149.1697,
134.3806, 138.2204, 146.3384, 117.8715, 152.3988, 172.9519, 159.7141, 132.7292,
137.4430, 190.0342, 116.9569, 134.5817, 149.3947, 130.4849, 124.4223, 127.3084,
108.9376, 151.0511]
```

```python
x_bar = sum(myDataset) / len(myDataset)
print(f"Stickprovsmedelvärde: {int(x_bar)}")
```

```python
mu = 100
print(f"Hypotetiskt populationsmedelvärde: {mu}")
```

```python
s = np.std(myDataset)
print(f"Standard deviation: {math.ceil(s)}")
```

```python
n = len(myDataset)
print(f"Stickprovsstorlek: {n}")
```

```python
from scipy.stats import t
# Tagen från ChatGPT
print("Utan inbyggd funktion \n")

def tTest():
    t_stat = (x_bar - mu) / (s / np.sqrt(n))
    df = n - 1
    p_value = 2 * (1 - t.cdf(abs(t_stat), df))
    print(f"T-statistic:: {t_stat:.2f}")
    print(f"P-value: {p_value}")
    return p_value
    
def checkPValue(p_value, alpha):
    if p_value < alpha:
        print(f"P-värdet är {p_value} vilket är mindre än {alpha}.")
        print("H0 förkastas: stickprovsmedelvärdet skiljer sig signifikant från populationsmedelvärdet.")
    else:
        print(f"P-värdet är {p_value} vilket är större än {alpha}.")
        print("H0 behålls: det finns ingen statistiskt signifikant skillnad mellan stickprovsmedelvärdet och populationsmedelvärdet.")
        
p = tTest()
print(checkPValue(p, alpha=0.05))
```

## 10 The One Sample t-test

### Hypoteserna

Vi jämför medelvärdet i ett stickprov (30 elever) med populationens medelvärde ($\mu = 4{,}2$).  
Därför formulerar vi följande hypoteser:

#### Nollhypotes ($H_0$):
$$
\mu = 4.2
$$
Medelvärdet i populationen är lika med 4,2.

#### Alternativ hypotes ($H_a$):
$$
\mu \neq 4.2
$$
Medelvärdet i populationen är inte lika med 4,2.

```python
# Tagen från stackoverflow

def mean_confidence_interval(data, confidence=0.95):
    a = 1.0 * np.array(data)
    n = len(a)
    m, se = np.mean(a), stats.sem(a)
    h = se * stats.t.ppf((1 + confidence) / 2., n-1)
    return m-h, m+h
```

```python
print(mean_confidence_interval(gradeData))
```

#### Alternativ hypotes ($H_a$):

## 12 Paired t-test

### Vad är behandlingen (treatment) i detta fall?

Behandlingen i detta fall är **undervisningssessionerna**.  
De två mätningarna (före och efter) har gjorts på samma elever, och det vi undersöker det med paired t-test är om undervisningen haft en signifikant effekt på testresultaten.

### Nollhypotes (H₀) och alternativ hypotes (Hₐ)

Vi undersöker om undervisningssessionerna har haft en effekt på studenternas testresultat.

#### **Nollhypotes (H₀):**
Det finns **ingen skillnad** i testresultat före och efter undervisningen.  
$$
\mu_{\text{före}} = \mu_{\text{efter}}
$$

#### **Alternativ hypotes (Hₐ):**
Det finns en **skillnad** i testresultat före och efter undervisningen.  
$$
\mu_{\text{före}} \neq \mu_{\text{efter}}
$$

Detta är ett **tvåsidigt test**, eftersom vi är intresserade av om det skett någon förändring, oavsett riktning (bättre eller sämre).
