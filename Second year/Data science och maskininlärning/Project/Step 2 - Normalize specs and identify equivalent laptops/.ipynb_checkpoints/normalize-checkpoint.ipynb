```python
import pandas as pd
import re

# Load both datasets
komplett_df = pd.read_excel('komplett_laptops_updated.xlsx')
elgiganten_df = pd.read_excel('elgiganten_laptops_updated.xlsx')

# Fill NaNs with empty string for processing
komplett_df.fillna('', inplace=True)
elgiganten_df.fillna('', inplace=True)

# --- Helper Functions ---
def extract_cpu(text):
    matches = re.findall(r'(Intel\s*Core\s*i\d[\w\-]*|AMD\s*Ryzen\s*\d[\w\-]*|Apple\s*M\d)', text, re.IGNORECASE)
    return matches[0] if matches else ''

def extract_ram(text):
    match = re.search(r'(\d+\s?GB)\s?RAM', text, re.IGNORECASE)
    return match.group(1) if match else ''

def extract_storage(text):
    match = re.search(r'(\d+\s?(TB|GB))\s?(SSD|eMMC|lagring)?', text, re.IGNORECASE)
    return match.group(1) if match else ''

def extract_gpu(text):
    match = re.search(r'(RTX\s*\d+|GTX\s*\d+|Radeon\s*\w+|Intel\s*Iris|Arc\s*Graphics)', text, re.IGNORECASE)
    return match.group(1) if match else ''

# --- Normalize Specs for Komplett ---
for i, row in komplett_df.iterrows():
    text = f"{row['Description']} {row['Name']}"
    if not row['CPU']:
        komplett_df.at[i, 'CPU'] = extract_cpu(text)
    if not row['RAM']:
        komplett_df.at[i, 'RAM'] = extract_ram(text)
    if not row['Storage']:
        komplett_df.at[i, 'Storage'] = extract_storage(text)
    if not row['GPU']:
        komplett_df.at[i, 'GPU'] = extract_gpu(text)

# --- Normalize Specs for Elgiganten ---
for i, row in elgiganten_df.iterrows():
    text = f"{row['Specs']} {row['Name']}"
    if not row['CPU']:
        elgiganten_df.at[i, 'CPU'] = extract_cpu(text)
    if not row['RAM']:
        elgiganten_df.at[i, 'RAM'] = extract_ram(text)
    if not row['Storage']:
        elgiganten_df.at[i, 'Storage'] = extract_storage(text)
    if not row['GPU']:
        elgiganten_df.at[i, 'GPU'] = extract_gpu(text)

# --- Harmonize column order ---
elgiganten_df = elgiganten_df[['Name', 'Price', 'CPU', 'RAM', 'Storage', 'GPU']]
komplett_df = komplett_df[['Name', 'Price', 'CPU', 'RAM', 'Storage', 'GPU']]

# --- Merge datasets ---
combined_df = pd.concat([komplett_df, elgiganten_df], ignore_index=True)

# --- Save to Excel ---
combined_df.to_excel("merged_laptops_cleaned.xlsx", index=False)
print("✅ Saved to merged_laptops_cleaned.xlsx")
```
