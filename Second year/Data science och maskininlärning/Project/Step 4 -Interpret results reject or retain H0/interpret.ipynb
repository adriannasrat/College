```python
import pandas as pd

results_df = pd.read_csv("../Step 3 - Group and match laptops using clustering or cosine similarity/ttest_results_by_cluster_1.csv", encoding="utf-8-sig")

for _, row in results_df.iterrows():
    cluster = int(row["Cluster"])
    p = row["P-Value"]
    if pd.isna(p):
        print(f"Cluster {cluster}: Not enough data for t-test.")
    elif p < 0.05:
        print(f"Cluster {cluster}: Reject H₀ → Significant price difference (p = {p:.10f})")
    else:
        print(f"Cluster {cluster}: Retain H₀ → No significant price difference (p = {p:.10f})")
```
