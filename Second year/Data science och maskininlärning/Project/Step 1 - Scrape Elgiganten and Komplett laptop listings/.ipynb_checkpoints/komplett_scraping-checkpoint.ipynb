```python
import asyncio
from playwright.async_api import async_playwright
import pandas as pd

def parse_specs(spec_string):
    cpu = ram = storage = screen = gpu = None
    specs = spec_string.split(",") if isinstance(spec_string, str) else []
    for spec in specs:
        spec_lower = spec.lower()
        if "ram" in spec_lower:
            ram = spec.strip()
        elif "ssd" in spec_lower or "lagring" in spec_lower or "emmc" in spec_lower:
            storage = spec.strip()
        elif "skärm" in spec_lower or '"' in spec_lower or "inch" in spec_lower or "tum" in spec_lower:
            screen = spec.strip()
        elif any(cpu_kw in spec for cpu_kw in ["intel", "amd", "apple"]):
            cpu = spec.strip()
        elif any(gpu_kw in spec for gpu_kw in ["rtx", "gtx", "radeon", "intel iris", "mx"]):
            gpu = spec.strip()
    return cpu, ram, storage, screen, gpu

async def scrape_komplett():
    async with async_playwright() as p:
        browser = await p.chromium.launch(headless=True, args=["--disable-http2"])
        context = await browser.new_context(
            user_agent="Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/122.0.0.0 Safari/537.36",
            viewport={"width": 1280, "height": 800}
        )

        page = await context.new_page()
        try:
            await page.goto(
                "https://www.komplett.se/category/10011/dator-surfplatta/datorer-barbara/laptop?list_view=list",
                timeout=60000,  # 60 seconds timeout
                wait_until="networkidle"
            )

            # Scroll in steps to mimic real user
            for _ in range(10):
                await page.mouse.wheel(0, 1000)
                await page.wait_for_timeout(1000)

            # Wait for at least one product box to appear
            await page.wait_for_selector("product-box", timeout=10000)

            boxes = await page.query_selector_all("product-box")
            data = []

            for box in boxes:
                try:
                    shadow = await box.evaluate_handle("el => el.shadowRoot")
                    a_tag = await shadow.query_selector("a.main-link")
                    if a_tag:
                        name = await a_tag.get_attribute("title")
                        specs = await a_tag.get_attribute("aria-label")
                        price = await a_tag.get_attribute("data-price")
                        url = await a_tag.get_attribute("href")
                        cpu, ram, storage, screen, gpu = parse_specs(specs)
                        data.append({
                            "Name": name,
                            "Price": price,
                            "Specs": specs,
                            "URL": url,
                            "CPU": cpu,
                            "RAM": ram,
                            "Storage": storage,
                            "Screen": screen,
                            "GPU": gpu
                        })
                except:
                    continue

            df = pd.DataFrame(data)
            df.to_excel("komplett_laptops_structured.xlsx", index=False)
            print(f"✅ Done! Saved {len(df)} laptops to 'komplett_laptops_structured.xlsx'")
        except Exception as e:
            print("❌ Failed to load page:", e)
        finally:
            await browser.close()

# Run this in Jupyter
await scrape_komplett()
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
import time
import pandas as pd

def parse_specs(spec_string):
    cpu = ram = storage = screen = gpu = None
    specs = spec_string.split(",") if isinstance(spec_string, str) else []
    for spec in specs:
        s = spec.lower()
        if "ram" in s:
            ram = spec.strip()
        elif any(x in s for x in ["ssd", "lagring", "emmc"]):
            storage = spec.strip()
        elif any(x in s for x in ['"', "inch", "tum", "skärm"]):
            screen = spec.strip()
        elif any(cpu_kw in spec for cpu_kw in ["intel", "amd", "apple"]):
            cpu = spec.strip()
        elif any(gpu_kw in spec for gpu_kw in ["rtx", "gtx", "radeon", "intel iris", "mx"]):
            gpu = spec.strip()
    return cpu, ram, storage, screen, gpu

# Setup browser
options = Options()
options.add_argument("--headless")
options.add_argument("--disable-gpu")
options.add_argument("--window-size=1920,1080")
driver = webdriver.Chrome(options=options)

print("🔄 Opening page...")
driver.get("https://www.komplett.se/category/10011/dator-surfplatta/datorer-barbara/laptop?list_view=list")
time.sleep(5)

# Scroll down
print("🔄 Scrolling...")
driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
time.sleep(5)

# Get product-boxes
product_boxes = driver.find_elements(By.CSS_SELECTOR, "product-box")
print(f"🔍 Found {len(product_boxes)} product-boxes")

data = []

for box in product_boxes:
    try:
        shadow = driver.execute_script("return arguments[0].shadowRoot", box)
        if not shadow:
            continue
        a_tag = shadow.find_element(By.CSS_SELECTOR, "a.main-link")
        name = a_tag.get_attribute("title")
        specs = a_tag.get_attribute("aria-label")
        price = a_tag.get_attribute("data-price")
        url = a_tag.get_attribute("href")
        cpu, ram, storage, screen, gpu = parse_specs(specs)
        data.append({
            "Name": name,
            "Price": price,
            "Specs": specs,
            "URL": url,
            "CPU": cpu,
            "RAM": ram,
            "Storage": storage,
            "Screen": screen,
            "GPU": gpu
        })
    except Exception as e:
        print("⚠️ Skipped one box:", e)

driver.quit()

# Save to Excel
df = pd.DataFrame(data)
df.to_excel("komplett_laptops_structured.xlsx", index=False)
print(f"✅ Done! Saved {len(df)} laptops to 'komplett_laptops_structured.xlsx'")
```

```python
import asyncio
from playwright.async_api import async_playwright
import csv

def parse_specs(spec_string):
    cpu = ram = storage = screen = gpu = None
    specs = spec_string.split(",") if isinstance(spec_string, str) else []
    for spec in specs:
        spec_lower = spec.lower()
        if "ram" in spec_lower:
            ram = spec.strip()
        elif "ssd" in spec_lower or "lagring" in spec_lower or "emmc" in spec_lower:
            storage = spec.strip()
        elif "skärm" in spec_lower or '"' in spec_lower or "inch" in spec_lower or "tum" in spec_lower:
            screen = spec.strip()
        elif any(cpu_kw in spec for cpu_kw in ["intel", "amd", "apple"]):
            cpu = spec.strip()
        elif any(gpu_kw in spec for gpu_kw in ["rtx", "gtx", "radeon", "intel iris", "mx"]):
            gpu = spec.strip()
    return cpu, ram, storage, screen, gpu

async def scrape_all_products():
    async with async_playwright() as p:
        browser = await p.chromium.launch(headless=False, slow_mo=100)
        context = await browser.new_context()
        page = await context.new_page()

        print("🔄 Opening page...")
        await page.goto("https://www.komplett.se/category/10011/dator-surfplatta/datorer-barbara/laptop?list_view=list", timeout=60000)

        # ⏳ Smart scroll loop
        last_count = 0
        max_stuck = 5
        stuck = 0

        while True:
            # Scroll to bottom using JS to trigger dynamic load
            await page.evaluate("window.scrollBy(0, document.body.scrollHeight)")
            await page.wait_for_timeout(2000)

            # Check product count
            boxes = await page.query_selector_all("product-box")
            current_count = len(boxes)
            print(f"📦 Boxes found: {current_count}")

            if current_count == last_count:
                stuck += 1
            else:
                stuck = 0
                last_count = current_count

            if stuck >= max_stuck:
                print(f"✅ Stopped scrolling after {current_count} boxes.")
                break

        # ✅ Scrape now
        data = []
        for i in range(current_count):
            try:
                selector = f"#product-box-{i}"
                box = await page.query_selector(selector)
                shadow = await box.evaluate_handle("el => el.shadowRoot")
                a_tag = await shadow.query_selector("a.main-link")

                name = await a_tag.get_attribute("title")
                specs = await a_tag.get_attribute("aria-label")
                price = await a_tag.get_attribute("data-price")
                url = await a_tag.get_attribute("href")

                cpu, ram, storage, screen, gpu = parse_specs(specs)

                data.append([name, price, specs, url, cpu, ram, storage, screen, gpu])
                print(f"✅ Scraped: {name}")
            except Exception as e:
                print(f"❌ Failed on box {i}: {e}")

        await browser.close()

        # 💾 Save it
        with open("komplett_laptops_structured.csv", mode="w", newline="", encoding="utf-8") as file:
            writer = csv.writer(file)
            writer.writerow(["Name", "Price", "Specs", "URL", "CPU", "RAM", "Storage", "Screen", "GPU"])
            writer.writerows(data)

        print(f"\n✅ DONE! Saved {len(data)} laptops to 'komplett_laptops_structured.csv'")

await scrape_all_products()
```

```python
import asyncio
from playwright.async_api import async_playwright
import csv

async def scrape_komplett_api():
    laptops = []

    async with async_playwright() as p:
        browser = await p.chromium.launch(headless=False)  # set to True if you don't want to see browser
        context = await browser.new_context()
        page = await context.new_page()

        # 🔁 Listen for API responses
        async def handle_response(response):
            if "GetSearchResult" in response.url and response.status == 200:
                try:
                    json_data = await response.json()
                    products = json_data.get("productListings", [])
                    for item in products:
                        laptops.append({
                            "Name": item.get("productTitle"),
                            "Price": item.get("productPricing", {}).get("discountedPrice", {}).get("amount"),
                            "Specs": item.get("productShortDescription", ""),
                            "URL": f"https://www.komplett.se{item.get('productPageUrl')}"
                        })
                        print(f"✅ {item.get('productTitle')}")
                except:
                    pass

        page.on("response", handle_response)

        print("🔄 Opening Komplett laptop page...")
        await page.goto("https://www.komplett.se/category/10011/dator-surfplatta/datorer-barbara/laptop?list_view=list")
        await page.wait_for_timeout(10000)  # wait for all XHRs

        await browser.close()

        # 💾 Save to CSV
        with open("komplett_laptops_from_api.csv", "w", newline="", encoding="utf-8") as file:
            writer = csv.writer(file)
            writer.writerow(["Name", "Price", "Specs", "URL"])
            writer.writerows([[l["Name"], l["Price"], l["Specs"], l["URL"]] for l in laptops])

        print(f"\n✅ DONE! Saved {len(laptops)} laptops to 'komplett_laptops_from_api.csv'")

# 🔁 Run this in your Jupyter or main.py
await scrape_komplett_api()
```

```python
import json
import pandas as pd

# --- Spec parser ---
def parse_specs(description, screen_size_text):
    cpu = ram = storage = screen = gpu = None
    specs = description.split(",") if isinstance(description, str) else []

    for spec in specs:
        spec_lower = spec.lower()
        if "ram" in spec_lower:
            ram = spec.strip()
        elif any(term in spec_lower for term in ["ssd", "lagring", "emmc"]):
            storage = spec.strip()
        elif any(term in spec_lower for term in ["intel", "amd", "apple", "core", "ryzen"]):
            cpu = spec.strip()
        elif any(term in spec_lower for term in ["rtx", "gtx", "radeon", "iris", "mx", "geforce", "arc"]):
            gpu = spec.strip()

    if screen_size_text:
        screen = screen_size_text.strip()

    return cpu, ram, storage, screen, gpu

# --- Load JSON ---
with open('data/komplett/komplett_page_12.json', 'r', encoding='utf-8') as file:
    data = json.load(file)

products = data.get('products', [])
product_list = []

for product in products:
    name = product.get('name', '')
    description = product.get('description', '')
    price = product.get('price', {}).get('listPrice', '')
    url = f"https://www.komplett.se{product.get('url', '')}"

    # Extract screen size if available in variantGroups
    screen_size_text = None
    for group in product.get('variantGroups', []):
        for attr in group.get('attributes', []):
            if attr.get('name') == "Skärmstorlek":
                screen_size_text = attr.get('value')
                break

    # Parse specs
    cpu, ram, storage, screen, gpu = parse_specs(description, screen_size_text)

    product_list.append({
        'Name': name,
        'Description': description,
        'Price': price,
        'URL': url,
        'CPU': cpu,
        'RAM': ram,
        'Storage': storage,
        'Screen': screen,
        'GPU': gpu
    })

# --- Export ---
df = pd.DataFrame(product_list)
df.to_excel('data/komplett/komplett_products_with_specs_12.xlsx', index=False)
print("✅ Data saved to 'komplett_products_with_specs.xlsx'")
```
