```python
import requests
from bs4 import BeautifulSoup
import csv 
```

```python
websites = [
     'https://www.techtarget.com/searchenterpriseai/definition/machine-learning-ML',
    'https://towardsdatascience.com/clustering-eating-behaviors-in-time-a-machine-learning-approach-to-preventive-health/']
 
output_file = 'scraped_products.txt'
```

```python
USER_AGENT = "Mozilla/5.0 (Macintosh; Intel Mac OS X 10.14; rv:65.0) Gecko/20100101 Firefox/65.0"
 
headers = {"user-agent": USER_AGENT}
r = requests.get("https://www.techtarget.com/searchenterpriseai/definition/machine-learning-ML", "https://towardsdatascience.com/",  headers=headers)
 
print("Status code: ", r.status_code)
```

```python
with open(output_file, 'w', encoding='utf-8') as file:
    for url in websites:
        response = requests.get(url, headers={'user-agent': 'Mozilla/5.0'})
        soup = BeautifulSoup(response.content, 'html.parser')
 
        title = soup.find('h1').get_text(strip=True) if soup.find('h1') else "No title"
        description = soup.find('p').get_text(strip=True) if soup.find('p') else "No description"
 
        file.write(f"Website: {url}\nTitle: {title}\nDescription: {description}\n\n---\n\n")
 
print(f"Scraping completed. Data saved in {output_file}.")
```

```python
USER_AGENT = "Mozilla/5.0 (Macintosh; Intel Mac OS X 10.14; rv:65.0) Gecko/20100101 Firefox/65.0"

headers = {"user-agent": USER_AGENT}
r = requests.get("https://www.elgiganten.se/datorer-kontor/datorer/laptop/windows-laptop", headers=headers)

print("Status code: ", r.status_code)
```

```python
if r.status_code == 200:
    soup = BeautifulSoup(r.content, "html.parser")
    products = soup.find_all("li", attrs={"data-sku": True})

    with open("elgiganten_laptops.csv", "w", newline="", encoding="utf-8-sig") as file:
        writer = csv.writer(file)
        writer.writerow(["Name", "Price", "SKU"])

        for product in products:
            # Try to extract name and price
            name_tag = product.find("h2") or product.find("h3")  # sometimes product name is in h2/h3
            price_tag = product.find("span", class_="price") or product.find("div", class_="product-price")
    
            name = name_tag.get_text(strip=True) if name_tag else "N/A"
            price = price_tag.get_text(strip=True) if price_tag else "N/A"
            sku = product.get("data-sku", "N/A")
    
            writer.writerow([name, price, sku])
    print("CSV file 'elgiganten_laptops.csv' created successfully.")

else:
    print("Scrape Error: status", r.status_code)
```
